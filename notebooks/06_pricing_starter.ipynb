## Appendix — Syntax toolbox (tools only)

Generic syntax on a **toy dataset** called `toy`, with made-up columns `y`, `x` and `group`. It does not touch `df`, and it does not tell you which variables to use: that is your group's decision. Copy a pattern, then replace the names.

Run A1 first. The toolbox keeps its data, models and tables in names that start with `toy` (`toy`, `toy_fit`, `toy_per_group`), so it does not overwrite your own `df` or models.

```python
# A1. Toy data, then a summary per group (pandas groupby)
N_PER_GROUP = 24                                  # rows per group
TOY_SLOPES = {'a': -0.8, 'b': -1.5, 'c': -2.2}    # true slope per group, to compare with A5
toy_rng = np.random.default_rng(0)
n_toy = N_PER_GROUP * len(TOY_SLOPES)
toy = pd.DataFrame({'group': np.repeat(list(TOY_SLOPES), N_PER_GROUP),
                    'x': toy_rng.uniform(5, 15, n_toy)})
toy['y'] = np.exp(5 + toy['group'].map(TOY_SLOPES) * np.log(toy['x']) + toy_rng.normal(0, 0.1, n_toy))

toy.groupby('group')[['x', 'y']].agg(['mean', 'min', 'max']).round(2)
```

```python
# A2. Log transform (only for strictly positive columns)
toy['log_y'] = np.log(toy['y'])
toy['log_x'] = np.log(toy['x'])
toy[['y', 'log_y', 'x', 'log_x']].describe().round(2)
```

```python
# A3. OLS with a formula. Pattern: 'y ~ x + C(group)'
#     C(...) turns a text column into one dummy per level; add more terms with +
toy_fit = smf.ols('log_y ~ log_x + C(group)', data=toy).fit()
print(toy_fit.summary().tables[1])      # coefficient table; print(toy_fit.summary()) shows everything
```

```python
# A4. Read one coefficient and its 95% confidence interval
b = toy_fit.params['log_x']
lo, hi = toy_fit.conf_int(alpha=0.05).loc['log_x']
print(f'coefficient on log_x: {b:.2f}  (95% CI {lo:.2f} to {hi:.2f}, p = {toy_fit.pvalues["log_x"]:.2g})')
print(f'R-squared: {toy_fit.rsquared:.3f} | observations: {int(toy_fit.nobs)}')
```

```python
# A5. One regression per group, results collected in a table
toy_rows = []
for g, d in toy.groupby('group'):
    toy_fit_g = smf.ols('log_y ~ log_x', data=d).fit()   # replace with your own formula
    lo, hi = toy_fit_g.conf_int().loc['log_x']
    toy_rows.append({'group': g, 'coef': toy_fit_g.params['log_x'], 'ci_low': lo, 'ci_high': hi, 'n': int(toy_fit_g.nobs)})
toy_per_group = pd.DataFrame(toy_rows).sort_values('coef')
toy_per_group.round(2)
```

```python
# A6. Revenue scenarios with constant elasticity: Q_new = Q_base * multiplier ** elasticity
def revenue_scenarios(price_base, qty_base, elasticity, multipliers=(0.9, 1.0, 1.1)):
    out = []
    for mult in multipliers:
        p = price_base * mult
        q = qty_base * mult ** elasticity
        out.append({'multiplier': mult, 'price': p, 'qty': q, 'revenue': p * q})
    return pd.DataFrame(out)

print(revenue_scenarios(price_base=10.0, qty_base=100.0, elasticity=-1.5).round(1))

# The same loop for every group (toy baseline: price 10, quantity 100 for all)
pd.concat({r.group: revenue_scenarios(10.0, 100.0, r.coef) for r in toy_per_group.itertuples()}).round(1)
```
